# Central tendency and dispersion of Key Metrics

- **Mean Sales and Profit** — The average value of transactions and their associated earnings.
- **Standard Deviation of Processing Days** — The variance in time taken to process and ship orders.

```python
# Importing  Py_Stats
import py_stats as ps


def descriptive_analytics(data):
    # Calculate measures of central tendency
    mean = ps.arithmetic_mean(data)
    median = ps.median(data)
    mode = ps.mode(data)
    # Calculate measures of dispersion
    variance = ps.variance(data,population=False)
    std_dev = ps.standard_deviation(data, population=False)
    range = ps.midrange(data)
    context = {
        "mean": mean,
        "median": median,
        "mode": mode,
        "variance": variance,
        "std_dev": std_dev,
        "range": range  
    }

    return context  
```

```python
sales = df['Sales'].to_list()
profit = df['Profit'].to_list()
processing_days = df['Order Processing Days'].to_list()
```

```python
sales_data = descriptive_analytics(sales)
profit_data = descriptive_analytics(profit)
processing_days_data = descriptive_analytics(processing_days)
```

```python
print("==================================\nDescriptive Analysis\n==================================")
print(f"Descriptive Analyis for Sales\n Mean : {sales_data['mean']:.2f}  \n Mode : {sales_data['mode']}\n Median : {sales_data['median']:.2f}\n Variance : {sales_data['variance']:.2f} \n Standard Deviation : {sales_data['std_dev']:.2f}\n Range : {sales_data['range']:.2f}\n==================================")
print(f"Descriptive Analyis for Profit\n Mean : {profit_data['mean']:.2f}  \n Mode : {profit_data['mode']}\n Median : {profit_data['median']:.2f}\n Variance : {profit_data['variance']:.2f} \n Standard Deviation : {profit_data['std_dev']:.2f}\n Range : {profit_data['range']:.2f}\n==================================")
print(f"Descriptive Analyis for Profit\n Mean : {processing_days_data['mean']:.2f}  \n Mode : {processing_days_data['mode']}\n Median : {processing_days_data['median']:.2f}\n Variance : {processing_days_data['variance']:.2f} \n Standard Deviation : {processing_days_data['std_dev']:.2f}\n Range : {processing_days_data['range']:.2f}\n==================================")
```

# Correlation Analysis
- **Sales vs. Profit Correlation** — Relationship strength between revenue levels and net earnings.
- **Discount vs. %Profit Margin Correlation** — Statistical impact of discount rates on overall profitability.

## Sales vs. Profit Correlation

```python
sales = df["Sales"].to_list()
profit = df["Profit"].to_list()

sales_profit_correlation = ps.pearson_correlation(sales,profit)
# 0.0<0.1 = No correlation 0.1<0.3 = little correlation 03<05 medium correlation 05<07 high correlation 0.7 < 1 very high correlation
if sales_profit_correlation == 0.0 or sales_profit_correlation <= 0.1:
    print("No Correlation")
elif sales_profit_correlation == 0.1 or sales_profit_correlation <= 0.3:
    print("Little Correlation")
elif sales_profit_correlation == 0.3 or sales_profit_correlation <= 0.5:
    print("Medium Correlation")
elif sales_profit_correlation == 0.5 or sales_profit_correlation <= 0.7:
    print("High Correlation")
else:
    print("Very High Correlation")
    
print(f"The Correlation between Sales and Profit : {sales_profit_correlation}")

```

## **Discount vs. %Profit Margin Correlation**

```python
discount = df["Discount"].to_list()
percentage_profit_margin = df["%Profit Margin"].to_list()

discount_percanatge_profit_margin_correlation = ps.pearson_correlation(discount,percentage_profit_margin)

if discount_percanatge_profit_margin_correlation == 0.0 or discount_percanatge_profit_margin_correlation <= 0.1:
    print("No Correlation")
elif discount_percanatge_profit_margin_correlation == 0.1 or discount_percanatge_profit_margin_correlation <= 0.3:
    print("Little Correlation")
elif discount_percanatge_profit_margin_correlation == 0.3 or discount_percanatge_profit_margin_correlation <= 0.5:
    print("Medium Correlation")
elif discount_percanatge_profit_margin_correlation == 0.5 or discount_percanatge_profit_margin_correlation <= 0.7:
    print("High Correlation")
else:
    print("Very High Correlation")
print(f"The Correlation between Discount and %Profit Margin : {discount_percanatge_profit_margin_correlation}")
```

# **Cohort Analysis**

### **Customer Retention by Order Year** 
— Analysis of repeat purchasing behavior based on the first Order Year.

```python
# make sure order dates are datetime and year column exists
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Order Year'] = df['Order Date'].dt.year

# first order year per customer
first_order = (df
               .groupby('Customer ID')['Order Year']
               .min()
               .rename('First Order Year'))
# attach to original dataframe
df = df.merge(first_order, on='Customer ID')

# count unique customers in each cohort year / activity year
cohort = (df
          .groupby(['First Order Year', 'Order Year'])['Customer ID']
          .nunique()
          .reset_index())

cohort_table = cohort.pivot(index='First Order Year',
                            columns='Order Year',
                            values='Customer ID')

# cohort size (customers who first ordered in that year)
cohort_sizes = cohort_table.iloc[:, 0]
# retention rate relative to cohort size
cohort_rate = cohort_table.div(cohort_sizes, axis=0)

print("Customer counts by cohort:")
display(cohort_table)

print("\nRetention rate (fraction of original cohort):")
display(cohort_rate)

# optional heatmap of retention rates
fig = px.imshow(cohort_rate,
                text_auto=True,
                labels={'x':'Order Year', 'y':'First Order Year', 'color':'Retention Rate'},
                title='Customer Retention by Order Year')
fig.show()
```

### **Monthly Order Cohorts** 
- Tracking customer activity based on their initial Number of Order Month.

```python
# derive first order month for each customer and merge it into df
first_order_month = (df
                     .groupby('Customer ID')['Number Of Order Month']
                     .min()
                     .rename('First Order Month'))
df = df.merge(first_order_month, on='Customer ID')

# build cohort counts for first‑order‑month vs. order‑month
cohort_month = (df
                .groupby(['First Order Month', 'Number Of Order Month'])['Customer ID']
                .nunique()
                .reset_index())

cohort_month_table = cohort_month.pivot(index='First Order Month',
                                        columns='Number Of Order Month',
                                        values='Customer ID')

# cohort sizes and retention rates
cohort_month_sizes = cohort_month_table.iloc[:, 0]
cohort_month_rate = cohort_month_table.div(cohort_month_sizes, axis=0)

print("Customer counts by monthly cohort:")
display(cohort_month_table)

print("\nRetention rate by monthly cohort:")
display(cohort_month_rate)

fig = px.imshow(cohort_month_rate,
                text_auto=True,
                labels={'x':'Order Month Number',
                        'y':'First Order Month Number',
                        'color':'Retention Rate'},
                title='Customer Retention by Order Month')
fig.show()
```
